## 3 DF from Dictionary

```python
import pandas as pd

# 1.Employee data
employees = {
    "emp_id": [1, 2, 3],
    "name": ["Ravi", "Priya", "Karthik"],
    "department": ["Sales", "IT", "HR"]
}
df_employees = pd.DataFrame(employees)

# 2.Sales data
sales = {
    "product": ["Laptop", "Mouse", "Keyboard"],
    "units_sold": [50, 200, 150],
    "price": [55000, 500, 1200]
}
df_sales = pd.DataFrame(sales)

# 3.Student marks
marks = {
    "student": ["A", "B", "C"],
    "math": [85, 90, 78],
    "science": [88, 76, 95]
}
df_marks = pd.DataFrame(marks)

print(df_employees)
print(df_sales)
print(df_marks)
```
